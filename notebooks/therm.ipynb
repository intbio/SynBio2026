```python
ppt_res = ppt_res.rename(columns={'EMB' : 'EMB2', 'EMB2' : 'EMB'})
ppt_res
```

```python
labels = pd.read_csv('../therm/labels.csv')
labels = labels.rename({'Functional Thermostability (C)': 'func_therm'}, axis = 1)
```

```python
labels["clas"] = (labels["func_therm"] > 72).astype(int)
```

```python
ppt_res['seq_len'] = ppt_res['Sequence'].apply(len)
```

```python
import Bio
from Bio.SeqUtils.ProtParam import ProteinAnalysis
import torch
DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'
DEVICE
```

```python
data = ppt_res
data
```

```python
def extract_descriptors(Sequence):
    analyzed_seq = ProteinAnalysis(Sequence)


    descriptors = {
        'molecular_weight': analyzed_seq.molecular_weight(),
        'aromaticity': analyzed_seq.aromaticity(),
        'instability_index': analyzed_seq.instability_index(),
        'isoelectric_point': analyzed_seq.isoelectric_point(),
        'gravy': analyzed_seq.gravy(),
        'charge_at_pH7': analyzed_seq.charge_at_pH(pH=7),
        'ss_a-helix': analyzed_seq.secondary_structure_fraction()[0],
        'ss_turn': analyzed_seq.secondary_structure_fraction()[1],
        'ss_b-sheet': analyzed_seq.secondary_structure_fraction()[2]
    }


    aa_composition = analyzed_seq.amino_acids_percent
    for aa, value in aa_composition.items():
        descriptors[f'aa_{aa}'] = value

    return descriptors

data['descriptors'] = data['Sequence'].apply(extract_descriptors)
descriptors_df = data['descriptors'].apply(pd.Series)
data = pd.concat([data, descriptors_df], axis=1)
data.head()
```

```python
data = data.iloc[1:].reset_index(drop=True)
```

```python
data
```

```python
data = data.drop('descriptors', axis=1)
```

```python

from transformers import AutoTokenizer, EsmModel
from tqdm import tqdm



tokenizer = AutoTokenizer.from_pretrained("facebook/esm2_t33_650M_UR50D")
model = EsmModel.from_pretrained("facebook/esm2_t33_650M_UR50D").to(DEVICE)
model.eval()

embeddings = []

for seq in tqdm(data["Sequence"]):
    inputs = tokenizer(seq, return_tensors="pt", truncation=True).to(DEVICE)

    with torch.no_grad():
        outputs = model(**inputs)

    emb = outputs.last_hidden_state[:, 1:-1].mean(dim=1).squeeze().cpu().numpy()
    embeddings.append(emb)
```

```python
X = pd.DataFrame(
    embeddings,
    columns=[f"emb_{i}" for i in range(len(embeddings[0]))]
)
data = pd.concat([data, X], axis=1)
```

```python
data_train = pd.read_csv('../therm/train.csv')
data_train
```

```python
x = data_train.drop(['seq_ID', 'Sequence', 'Unnamed: 0'], axis = 1)
x
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, labels.clas, test_size = 0.25, stratify = labels.clas)
print(x_train.shape, x_test.shape, y_train.shape, y_test.shape)
```

```python
from sklearn.ensemble import GradientBoostingClassifier
model_cl_grad = GradientBoostingClassifier(n_estimators=150)
model_cl_grad.fit(x_train, y_train)
y_pred_grad = model_cl_grad.predict(x_test)
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred_grad))
```

```python
from sklearn.ensemble import ExtraTreesClassifier
model_cl_extra = ExtraTreesClassifier(n_estimators=1500)
model_cl_extra.fit(x_train, y_train)
y_pred_extra = model_cl_extra.predict(x_test)

print(classification_report(y_test, y_pred_extra))
```
